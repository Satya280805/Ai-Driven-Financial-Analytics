```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/processed/cleaned_financial_data.csv")
df.head()
```

```python
df.info()
```

```python
df["Transaction_Date"] = pd.to_datetime(
    df["Transaction_Date"],
    format="mixed",
    dayfirst=True
)
```

```python
df["Year"] = df["Transaction_Date"].dt.year
```

```python
df["Month"] = df["Transaction_Date"].dt.month
```

```python
df["Quarter"] = df["Transaction_Date"].dt.quarter
```

```python
df[["Transaction_Date", "Year", "Month", "Quarter"]].head(10)
```

```python
df[["Year", "Month", "Quarter"]].nunique()
```

```python
df["Budget_Variance"] = df["Amount"] - df["Budget"]
```

```python
df[["Amount", "Budget", "Budget_Variance"]].head(10)
```

```python
df["Budget_Utilization"] = (
    df["Amount"] / df["Budget"]
) * 100
```

```python
df[["Amount", "Budget", "Budget_Utilization"]].head(10)
```
