Cleaning

```python
account_category_mapping = {
    "Office_Supplies": "Office Supplies",
    "logistics": "Logistics",
    "SALES": "Sales",
    "maintenance": "Maintenance",
    "Marketting": "Marketing",
    "utilities": "Utilities",
    "TRAVEL": "Travel",
    "I.T.": "IT",
    "salaries": "Salaries",
    "other": "Other"
}

df["Account_Category"] = df["Account_Category"].replace(account_category_mapping)

df["Department"] = df["Department"].str.strip()

df["Company_Code"] = df["Company_Code"].str.strip().str.upper()
```

```python
print("Account Categories:")
print(df["Account_Category"].unique())

print("\nDepartments:")
print(df["Department"].unique())

print("\nCompany Codes:")
print(df["Company_Code"].unique())
```

```python
df.to_csv("../data/processed/cleaned_financial_data.csv", index=False)
```
